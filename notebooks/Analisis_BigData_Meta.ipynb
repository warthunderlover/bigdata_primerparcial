# Análisis de Big Data — Meta Platforms
**Clase:** Big Data | **Carrera:** Ingeniería en Ciencias de la Computación | **Año:** 2026

Elegí Meta porque es básicamente el ejemplo más claro de una empresa que no podría existir sin Big Data. Todo su negocio, desde los anuncios hasta el feed, depende de procesar datos a una escala enorme. En 2024 tuvieron ingresos de $164.5 mil millones y el 97.5% vino de publicidad, que funciona gracias al análisis de datos de sus usuarios.

---
## Introducción

Meta opera Facebook, Instagram, WhatsApp y Messenger. Cada acción que hace un usuario, ya sea un like, un mensaje o ver un video, genera datos que Meta guarda y analiza.

| Plataforma | Usuarios activos diarios |
|---|---|
| Facebook | 2,110 millones |
| Instagram | ~1,000 millones |
| WhatsApp | ~1,000 millones |
| Messenger | ~394 millones |
| **Total** | **3,580 millones** |

---
## V1 — Volumen

Meta genera alrededor de 4 petabytes de datos al día solo en Facebook, almacenados en su sistema interno llamado Hive. Un petabyte equivale a un millón de gigabytes, así que estamos hablando de una cantidad de datos que es prácticamente imposible de imaginar.

| Actividad | Por minuto | Por día |
|---|---|---|
| Comentarios en Facebook | 510,000 | 734 millones |
| Fotos subidas | 136,000 | 195 millones |
| Likes | 4,000,000 | 5,760 millones |
| Reels reproducidos (FB + IG) | 138,900,000 | 200 mil millones |
| Mensajes en WhatsApp | 69,000,000 | 100 mil millones |

En total el ecosistema completo de Meta genera aproximadamente **8.4 petabytes por día**.

---
## V2 — Velocidad

Todos esos datos no sirven de nada si no se procesan rápido. El algoritmo del feed se actualiza en milisegundos con cada interacción, el sistema de publicidad decide qué anuncio mostrar en menos de 100ms, y WhatsApp entrega mensajes en menos de un segundo a cualquier parte del mundo.

| Evento | Por segundo |
|---|---|
| Reproducciones de Reels | ~2,315,000 |
| Mensajes en WhatsApp | ~1,150,000 |
| Likes | ~66,667 |
| Comentarios | ~8,500 |
| Fotos subidas | ~2,267 |

Para manejar esto usan tecnologías como Apache Spark Streaming y Kafka, que están diseñadas específicamente para procesar flujos de datos en tiempo real.

---
## V3 — Variedad

Meta no solo maneja un tipo de dato, sino prácticamente todos los que existen al mismo tiempo.

**Estructurados** (tienen formato fijo): datos de perfil como edad, ciudad e idioma; historial de anuncios; datos del dispositivo; transacciones en Marketplace.

**Semiestructurados** (formato parcial, como JSON): logs de clics y tiempo por post, metadatos de archivos, conexiones entre usuarios.

**No estructurados** (sin formato definido): texto libre de publicaciones y comentarios, las 195 millones de fotos diarias, videos, notas de voz y llamadas de WhatsApp.

La combinación de estos tres tipos es uno de los principales desafíos técnicos de Meta, porque cada uno requiere tecnologías de almacenamiento y procesamiento distintas.

---
## V4 — Veracidad

No todos los datos que genera Meta son confiables. Este es quizás el problema más difícil de resolver porque afecta directamente la calidad de todo lo demás.

| Problema | Escala |
|---|---|
| Cuentas falsas eliminadas | +1,000 millones por trimestre |
| Spam y bots | Millones de interacciones diarias |
| Artículos de desinformación | Miles por día |
| Ubicaciones GPS erróneas | Millones por día |

Para limpiar esto usan machine learning que detecta comportamiento anómalo, computer vision para identificar imágenes manipuladas, y pipelines automáticos de validación de datos antes de que lleguen a los modelos de IA. Si no filtraran esto, los algoritmos de publicidad estarían entrenados con basura y el negocio se vendría abajo.

## Pregunta 1 — ¿Qué datos genera?

Meta genera datos de casi cada acción que hacen sus 3,580 millones de usuarios diarios. El tipo varía mucho: desde texto y fotos hasta datos de comportamiento como cuánto tiempo pasaste mirando un post antes de seguir haciendo scroll, o con qué batería tenía el teléfono cuando abriste la app.

| Tipo | Ejemplos | Frecuencia |
|---|---|---|
| Texto | Posts, comentarios, mensajes | Continua |
| Imágenes | Fotos, Stories | 136,000/minuto |
| Video | Reels, Lives | 138.9M reproduc./minuto |
| Comportamiento | Scroll, clics, tiempo por post | Cada milisegundo |
| Ubicación | GPS, ciudad, IP | Por sesión |
| Dispositivo | Modelo, OS, batería | Por sesión |

---
## Pregunta 2 — ¿Quién usa estos datos?

Los datos no los usa solo Meta. Hay varios actores involucrados:

**Meta internamente** los usa para entrenar sus algoritmos, personalizar el feed y desarrollar modelos de IA como LLaMA.

**Los anunciantes**, que son más de 10 millones activos, pagan para llegar a audiencias segmentadas por edad, intereses y comportamiento.

**Los gobiernos** hacen solicitudes legales de datos de usuarios en investigaciones criminales. Esto ocurre miles de veces por semestre a nivel mundial.

**Investigadores académicos** usan datos anonimizados para estudiar comportamiento social y propagación de desinformación.

**Data Brokers** combinan datos públicos de Meta con otras fuentes para crear perfiles demográficos que venden a empresas de marketing y seguros.

---
## Pregunta 3 — ¿Qué problema resuelve?

El análisis de datos en Meta resuelve problemas en varios niveles.

Para la empresa, el problema principal es cómo mostrarle a cada uno de los 3,580 millones de usuarios contenido y publicidad que sea relevante para él. Sin análisis de datos esto sería imposible porque no habría forma de saber qué le interesa a cada persona.

Para la sociedad, los datos han permitido detectar redes de desinformación en períodos electorales, colaborar con gobiernos en crisis de salud pública, y detectar abuso infantil antes de que se propague.

Para los usuarios, el valor más directo es la personalización del contenido y la traducción automática en más de 100 idiomas, que permite que personas de diferentes países se comuniquen sin barreras.

---
## Pregunta 4 — ¿Por qué no bastaría Excel?

Excel tiene un límite de 1,048,576 filas. Meta recibe 734 millones de comentarios solo en Facebook en un día, lo que equivale a necesitar 700 archivos de Excel para guardar únicamente los comentarios de un día de una sola plataforma. Eso ya lo descarta por volumen.

Pero el problema no es solo el tamaño. Excel es estático y manual. Meta necesita decidir en menos de 100 milisegundos qué anuncio mostrarle a cada usuario cuando carga el feed. Eso requiere sistemas de procesamiento en tiempo real como Apache Kafka y Spark Streaming, que no tienen nada que ver con una hoja de cálculo.

Además, Excel solo maneja texto y números en celdas. No puede almacenar videos, grafos de relaciones sociales ni datos de comportamiento en tiempo real. Para eso Meta usa bases de datos NoSQL como Cassandra, sistemas de archivos distribuidos como HDFS, y bases de datos de grafos.

En resumen, el problema no es solo que Excel sea pequeño, sino que es una herramienta diseñada para un tipo de trabajo completamente diferente al que necesita Meta.

---
## Conclusión

Meta es un buen caso de estudio porque no es una empresa que usa Big Data como herramienta adicional, sino que es Big Data. Sin la capacidad de procesar petabytes en tiempo real, personalizar contenido para 3,580 millones de personas y monetizar cada interacción, Meta no existiría.

Las tecnologías que usa, Hadoop, Spark, Kafka y Cassandra, son exactamente las que se estudian en este curso, lo que hace que entender cómo las aplica Meta ayude a comprender para qué sirve cada una en la práctica.
